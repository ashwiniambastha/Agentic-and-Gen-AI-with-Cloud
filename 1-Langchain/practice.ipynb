```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-120b")   
```

```python
llm.invoke("write 50 words about me?")
```

```python
from langchain_core.prompts import ChatPromptTemplate


prompt=ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You are AI expert but you only reply in 40 words of every question"),
        ("user", "{input}")
    ]
)

```

```python
prompt
```

```python
ai= prompt|llm
```

```python
ai.invoke("what is rag?")
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()


ai=prompt|llm|parser
```

```python
ai.invoke("what is rag?")
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

parser=JsonOutputParser()

prompt=PromptTemplate(
    template="You are AI expert but you only reply in 40 words of every question. {input}{format_instructions}",
    input_varibles=["input"],
    partial_variables={"format_instructions": parser.get_format_instructions()}
)
```

```python
ai=prompt|llm|parser
```

```python
ai.invoke("what is rag?")
```

# Document Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader=WebBaseLoader("https://www.geeksforgeeks.org//")
```

```python
docs=loader.load()
```

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=75)
```

```python
document=text_splitter.split_documents(docs)
```

```python
document
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("nihabilal/qwen3-jailbreak-embedding-model")

```

```python
from langchain_community.vectorstores import FAISS

vectorstore=FAISS.from_documents(documents,model)
vectorstore
```
