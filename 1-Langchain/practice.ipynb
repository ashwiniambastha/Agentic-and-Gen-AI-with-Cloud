```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

## Langchain Tracking and tracing

os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-120b")   
```

```python
llm.invoke("write 50 words about me?")
```

```python
llm.invoke("write 50 words about me?").content
```

```python
from langchain_core.prompts import ChatPromptTemplate


prompt=ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You are AI expert but you only reply in 40 words of every question"),
        ("user", "{input}")
    ]
)

```

```python
prompt
```

```python
ai= prompt|llm
```

```python
ai.invoke("what is rag?")
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()


ai=prompt|llm|parser
```

```python
ai.invoke("what is rag?")
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

parser=JsonOutputParser()

prompt=PromptTemplate(
    template="You are AI expert but you only reply in 40 words of every question. {input}{format_instructions}",
    input_varibles=["input"],
    partial_variables={"format_instructions": parser.get_format_instructions()}
)
```

```python
ai=prompt|llm|parser
```
